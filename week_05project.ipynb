```python
#import necessary functions
from IPython.display import clear_output
# global list variable
cart=[]
```

```python
#create a function to add items to a cart
def addItem(item):
    clear_output()
    cart.append(item)
    print("{} has been added.".format(item))
```

```python
#create function to remove items from cart
def removeItem(item):
    clear_output()
    try:
        cart.remove(item)
        print("{} has been removed .".format(item))
    except:
        print("Sorry we could not remove that item.")
```

```python
# create a function to show items in cart
def showCart():
    clear_output()
    if cart:
        print("Here is your cart:")
    for item in cart:
        print("-{}".format(item))
    else:
        print("Your cart is empty.")
```

```python
# create function to clear items from cart
def clearCart():
    clear_output()
    cart.clear()
    print("Your cart is empty.")
```

```python
# create main function that loops until user quits
def main():
    done=False
    while not done:
        ans=input("quit/add/remove/show/clear:").lower()
        #base case
        if ans=="quit":
            print("Thanks for using our program.")
            showCart()
            done=True
main() #run the program
```
